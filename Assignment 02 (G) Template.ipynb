Task 02: Which Seniors Use AI Most for Assignments?
</h3>

<p>
<strong>Question:</strong>
Which senior students reported using AI on assignments?
Display their <code>Major</code>, <code>AI_Tool</code>,
and <code>AI_Hours_Per_Week</code> for students who
answered <code>Yes</code> to <code>Used_AI_Assignment</code>.
Arrange the results so that the heaviest AI users appear first.

```python
# Check exact column names in your DataFrame
print(df_ai.columns.tolist())
```

```python
import sqlite3
import pandas as pd

# 1. Load the CSV file
df_ai = pd.read_csv("Course_Evaluation.csv")

# 2. Connect to SQLite database and write table
conn = sqlite3.connect("ai_classroom.db")
df_ai.to_sql("course_eval", conn, if_exists="replace", index=False)

# 3. SQL Query using 'Academic_Standing'
query_task2 = """
SELECT Major, AI_Tool, AI_Hours_Per_Week
FROM course_eval
WHERE Academic_Standing = 'Senior' AND Used_AI_Assignment = 'Yes'
ORDER BY AI_Hours_Per_Week DESC;
"""

# 4. Execute query and display results
df_task2 = pd.read_sql_query(query_task2, conn)

print("--- Seniors Using AI for Assignments (Sorted by Usage Hours) ---")
print(df_task2)

# Close database connection
conn.close()
```

Task 03: Which AI Tools Are Most Commonly Reported?
</h3>

<p>
<strong>Question:</strong>
Which <code>AI_Tool</code> entries appear most frequently
in the survey? Display the number of students reporting
each tool entry, arranged from most to least common.
Identify the most frequently reported entry.
</p>

</div>

```python
# Query frequency of each AI_Tool entry from most to least common
query_task3 = """
SELECT AI_Tool, COUNT(*) AS Student_Count
FROM course_eval
GROUP BY AI_Tool
ORDER BY Student_Count DESC;
"""

df_task3 = pd.read_sql_query(query_task3, conn)
print("--- AI Tool Usage Frequency ---")
print(df_task3)

# Identify the most frequently reported entry
top_tool = df_task3.iloc[0]["AI_Tool"]
top_count = df_task3.iloc[0]["Student_Count"]
print(
    f"\nMost Frequently Reported Tool Entry: {top_tool} ({top_count} students)"
)
```

Task 04: Does AI Proficiency Relate to Learning Impact?
</h3>

<p>
<strong>Question:</strong>
How does the average <code>AI_Learning_Impact</code>
differ across <code>AI_Proficiency</code> levels?
For each proficiency level, report the number of students
and the average learning impact rounded to two decimal places.
Identify the proficiency level with the highest average
learning impact.
</p>

</div>

```python
# Calculate average AI_Learning_Impact and count by AI_Proficiency level
query_task4 = """
SELECT 
    AI_Proficiency,
    COUNT(*) AS Student_Count,
    ROUND(AVG(AI_Learning_Impact), 2) AS Avg_Learning_Impact
FROM course_eval
GROUP BY AI_Proficiency
ORDER BY Avg_Learning_Impact DESC;
"""

df_task4 = pd.read_sql_query(query_task4, conn)
print("--- Learning Impact by AI Proficiency Level ---")
print(df_task4)

# Identify the proficiency level with highest learning impact
highest_impact_level = df_task4.iloc[0]["AI_Proficiency"]
highest_impact_score = df_task4.iloc[0]["Avg_Learning_Impact"]
print(
    f"\nHighest Average Learning Impact Level: {highest_impact_level} (Score: {highest_impact_score})"
)
```

Task 05: Which Frequency Groups Spend More Time Using AI?
</h3>

<p>
<strong>Question:</strong>
Which <code>AI_Usage_Frequency</code> groups average
more than one hour per week using AI?
Display each qualifying group and its average
<code>AI_Hours_Per_Week</code>.
Include only groups whose average exceeds one hour.

```python
# Query frequency groups with average AI hours greater than 1 hour per week
query_task5 = """
SELECT 
    AI_Usage_Frequency,
    ROUND(AVG(AI_Hours_Per_Week), 2) AS Avg_Hours_Per_Week
FROM course_eval
GROUP BY AI_Usage_Frequency
HAVING AVG(AI_Hours_Per_Week) > 1.0
ORDER BY Avg_Hours_Per_Week DESC;
"""

df_task5 = pd.read_sql_query(query_task5, conn)
print("--- Usage Frequency Groups Averaging > 1 Hour/Week ---")
print(df_task5)

# Close database connection when finished
conn.close()
```

```python
# 1. Calculate average SAT_ACT score for each college
avg_sat_by_college = df_admin.groupby("College")["SAT_ACT"].mean()
print("--- Average SAT_ACT Score by College ---")
print(avg_sat_by_college.round(2))

# 2. Calculate overall average test score across all applicants
overall_avg_sat = df_admin["SAT_ACT"].mean()
print(f"\nOverall Average SAT_ACT Score: {overall_avg_sat:.2f}")

# 3. Identify the college with the highest average score
top_score_college = avg_sat_by_college.idxmax()
top_score_val = avg_sat_by_college.max()

print(
    f"College with Highest Average Test Score: {top_score_college} ({top_score_val:.2f})"
)
```
